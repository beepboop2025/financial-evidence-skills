# China Economy Revisions & Information-Control Verification (Palimpsest)

Fetch and cryptographically fingerprint the latest public reading from [Palimpsest](https://www.palimpsest.info/china/), an open-source, revision-safe China economic observatory and verifiable AI/censorship measurement registry.

- China observatory: https://www.palimpsest.info/china/
- Public reading JSON: https://www.palimpsest.info/readings/china-index-latest.json
- Source repository: https://github.com/beepboop2025/palimpsest

**Boundary:** Public open-source research observations. Separate observation clocks, publisher release times, and retrieval times; never substitute missing readings with zero.

```python
import hashlib
import json
from datetime import datetime, timezone
from urllib.request import Request, urlopen

URL = "https://www.palimpsest.info/readings/china-index-latest.json"
req = Request(URL, headers={"Accept": "application/json", "User-Agent": "Palimpsest-Colab-Verify/1.0"})
with urlopen(req, timeout=20) as resp:
    raw = resp.read()

digest = hashlib.sha256(raw).hexdigest()
data = json.loads(raw.decode("utf-8"))
print("Retrieved UTC:", datetime.now(timezone.utc).isoformat())
print("Bytes:", len(raw), "SHA-256:", digest)
print("Top-level keys:", sorted(data.keys()))
print(json.dumps(data, indent=2)[:2000])
```
